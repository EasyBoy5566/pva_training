# 依模式擷取呼吸

載入錄製檔 → 選擇要輸出的模式 → 輸出 summary 與 waves 兩個 CSV。

只需要錄製檔（`.csv` / `.csv.gz`），不需要 labels 檔。

```python
import os
import numpy as np
import pandas as pd

WAVE_COLS = ["Time (s)", "Pressure (cmH2O)", "Flow (L/min)", "Volume (mL)"]
```

## 函式

直接執行即可，不需修改。

```python
def load_recording(path):
    """載入錄製檔（.csv / .csv.gz）。Mode 去前後空白；Event 欄舊格式沒有就補空。"""
    df = pd.read_csv(path, dtype={"Mode": str, "Event": str}, low_memory=False,
                     usecols=lambda c: c in WAVE_COLS + ["Trigger", "Mode", "Event"])
    df["Mode"] = df["Mode"].str.strip().replace("", np.nan)
    if "Event" not in df:
        df["Event"] = np.nan
    return df


def list_modes(df):
    """錄製檔裡出現過的模式（依出現順序）"""
    return list(dict.fromkeys(df["Mode"].dropna()))


def _mode_segments(df, latency=1.5):
    """每段模式的 [start_time, end_time)；模式切換點對齊到 trigger（同你原本的邏輯）"""
    t = df["Time (s)"].to_numpy()
    trig_t = t[df["Trigger"].to_numpy() == 1]
    polls = df.dropna(subset=["Mode"])
    pt, pm = polls["Time (s)"].to_numpy(), polls["Mode"].to_numpy()
    if len(pm) == 0:
        return pd.DataFrame(columns=["mode", "start_time", "end_time"])

    starts, modes = [t[0]], [pm[0]]
    for k in np.flatnonzero(pm[1:] != pm[:-1]) + 1:
        tA, tB = pt[k - 1], pt[k]
        new = trig_t[(trig_t > tA) & (trig_t <= tB)]
        old = trig_t[(trig_t <= tA) & (trig_t >= tA - latency)]
        b = new[-1] if len(new) else (old[-1] if len(old) else tA)
        starts.append(max(b, starts[-1]))
        modes.append(pm[k])

    segs = pd.DataFrame({"mode": modes, "start_time": starts})
    segs["end_time"] = segs["start_time"].shift(-1).fillna(t[-1] + 0.01)
    return segs[segs["end_time"] > segs["start_time"]].reset_index(drop=True)


def _flow_starts(t, f, pos=2.0, neg=-2.0, min_breath=1.0):
    """沒有 trigger 時：flow 先低於 neg，再向上穿越 pos，起點取緊鄰的負轉正零交越點"""
    starts, armed, last = [], False, -np.inf
    for i in range(1, len(f)):
        if f[i] <= neg:
            armed = True
        elif armed and f[i - 1] < pos <= f[i]:
            j = i
            while j > 0 and f[j - 1] >= 0:
                j -= 1
            if t[j] - last >= min_breath:
                starts.append(t[j]); last = t[j]; armed = False
    return np.array(starts)


def extract_mode(rec_path, modes, out_dir=None, split="auto", latency=1.5):
    """
    從錄製檔取出指定模式的每一口呼吸並輸出。

    rec_path : 錄製檔路徑
    modes    : 要的模式，字串或 list，例如 "PC-PSV" 或 ["PC-PSV", "SPN-CPAP"]
    out_dir  : 輸出資料夾；None 則不寫檔，只回傳 DataFrame
    split    : "trigger" 用 Trigger 欄切；"flow" 用 flow 負轉正切；
               "auto" 每段模式內 trigger 少於 2 個就改用 flow
    回傳 (summary, waves) 兩個 DataFrame
    """
    if isinstance(modes, str):
        modes = [modes]
    modes = [m.strip() for m in modes]

    df = load_recording(rec_path)
    available = list_modes(df)
    missing = [m for m in modes if m not in available]
    if missing:
        print(f"找不到模式 {missing}，此檔有的模式：{available}")

    # 斷線/重連標記列（波形欄空白）先記下時間再拿掉
    gap_t = df.loc[df["Event"].notna(), "Time (s)"].to_numpy()
    segs = _mode_segments(df, latency)
    wave = df.dropna(subset=["Pressure (cmH2O)"]).reset_index(drop=True)
    t = wave["Time (s)"].to_numpy()
    trig = wave["Trigger"].to_numpy() == 1

    rows = []
    for seg_id, s in segs[segs["mode"].isin(modes)].iterrows():
        a = np.searchsorted(t, s.start_time)
        b = np.searchsorted(t, s.end_time)
        st = t[a:b][trig[a:b]]
        source = "trigger"
        if split == "flow" or (split == "auto" and len(st) < 2):
            st = _flow_starts(t[a:b], wave["Flow (L/min)"].to_numpy()[a:b])
            source = "flow"
        # 一口 = [起點_k, 起點_k+1)；段內最後一個起點之後不完整，捨棄
        for s0, s1 in zip(st[:-1], st[1:]):
            if ((gap_t >= s0) & (gap_t < s1)).any():
                continue                      # 跨斷線的那口不要
            rows.append((s["mode"], seg_id, source, s0, s1))

    summary = pd.DataFrame(rows, columns=["mode", "segment_id", "split_by", "start_time", "end_time"])
    summary.insert(0, "breath_id", np.arange(1, len(summary) + 1))

    pieces = []
    for r in summary.itertuples():
        a = np.searchsorted(t, r.start_time)
        b = np.searchsorted(t, r.end_time)       # 不含下一口的起點
        w = wave.iloc[a:b][WAVE_COLS].copy()
        w.insert(0, "t_rel (s)", (w["Time (s)"] - r.start_time).round(3))
        w.insert(0, "mode", r.mode)
        w.insert(0, "breath_id", r.breath_id)
        pieces.append(w)
    waves = pd.concat(pieces, ignore_index=True) if pieces else pd.DataFrame(
        columns=["breath_id", "mode", "t_rel (s)"] + WAVE_COLS)

    g = waves.groupby("breath_id")
    summary["duration (s)"] = (summary["end_time"] - summary["start_time"]).round(3)
    summary["n_samples"] = summary["breath_id"].map(g.size()).fillna(0).astype(int)
    summary["peak_pressure"] = summary["breath_id"].map(g["Pressure (cmH2O)"].max())
    summary["peak_flow"] = summary["breath_id"].map(g["Flow (L/min)"].max())
    summary["max_volume"] = summary["breath_id"].map(g["Volume (mL)"].max())

    if out_dir:
        os.makedirs(out_dir, exist_ok=True)
        base = os.path.basename(rec_path).split(".")[0]
        tag = "_".join(m.replace("/", "-") for m in modes)
        summary.to_csv(os.path.join(out_dir, f"{base}_{tag}_summary.csv"), index=False, encoding="utf-8-sig")
        waves.to_csv(os.path.join(out_dir, f"{base}_{tag}_waves.csv"), index=False, encoding="utf-8-sig")

    print(f"{len(summary)} 口呼吸（{', '.join(modes)}）")
    return summary, waves
```

## 1. 設定錄製檔

```python
REC = "2223585_20260712_0508.csv.gz"   # 錄製檔路徑
OUT_DIR = "output"                       # 輸出資料夾；設 None 則不寫檔
```

## 2. 查看這個檔有哪些模式

```python
df = load_recording(REC)
list_modes(df)
```

## 3. 選擇模式並輸出

- `MODES`：字串或 list，例如 `"PC-PSV"` 或 `["PC-PSV", "SPN-CPAP"]`
- `SPLIT`：`"auto"`（預設，段內 trigger 少於 2 個改用 flow）、`"trigger"`、`"flow"`

```python
MODES = "PC-PSV"
SPLIT = "auto"

summary, waves = extract_mode(REC, MODES, out_dir=OUT_DIR, split=SPLIT)
```

## 4. 檢視結果
